```python
from diffusers import DiffusionPipeline

pipe = DiffusionPipeline.from_pretrained("maxmarcon/klimt-diffusion").to("cuda")
```

```python
images = pipe(10)
```

```python
for i, im in enumerate(images[0]):
    im.save(f'{i}.jpg')


```
